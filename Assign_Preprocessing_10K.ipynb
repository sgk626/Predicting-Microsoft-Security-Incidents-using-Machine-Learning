### Preprocessing - step 1
#### The original data from Kaggle split data into Test and Train data. The step here is to consolidate both files into 1 file and extract 1% of datasets for this assignment. The reason that not full dataset is used due to low laptop capability to compute millions of data records.

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
# Importing the dataset
df_train = pd.read_csv('D:/sgk62/Documents/Master of AI/AML/Assignment Data/Microsoft Security Incident Prediction/GUIDE_Train.csv')
df_test = pd.read_csv('D:/sgk62/Documents/Master of AI/AML/Assignment Data/Microsoft Security Incident Prediction/GUIDE_Test.csv')
```

```python
# Combine 2 files into 1 file
df_combined = pd.concat([df_train, df_test], ignore_index=True)
```

```python
# Extract 1% of the DataFrame
df = df_combined.sample(frac=0.001, random_state=1)

# Define the path where you want to save the file
file_path = r'D:\sgk62\Documents\Master of AI\AML\Assignment Data\Microsoft Security Incident Prediction\DF_Sample_10K.csv'

# Save the sampled DataFrame to the specified path
df.to_csv(file_path, index=False)

print(f"Sampled DataFrame saved to {file_path}")
```
